```python
import pandas as pd
from sqlalchemy import create_engine
import urllib.parse

# 1. Re-read your original Kaggle CSV file into Python
# (Replace 'your_kaggle_file.csv' with your actual file path)
df = pd.read_csv(r"C:\Users\Vinay\Downloads\customer_shopping_behavior.csv")

# 2. THE PERMANENT CLEAN: This fixes spaces AND removes parentheses from ALL columns instantly
df.columns = df.columns.str.replace(' ', '_')      # Replaces spaces with _
df.columns = df.columns.str.replace('(', '')       # Removes opening brackets
df.columns = df.columns.str.replace(')', '')       # Removes closing brackets

# Let's print them to verify they look beautiful (e.g., Purchase_Amount_USD)
print("Perfectly cleaned columns:", df.columns.tolist())

# 3. Connect to your MySQL database
USER = "root"             
PASSWORD = "Ksasi@2003#jmp"  
HOST = "localhost"         
PORT = "3306"              
DB_NAME = "customer_behaviour"  

safe_password = urllib.parse.quote_plus(PASSWORD)
connection_string = f"mysql+pymysql://{USER}:{safe_password}@{HOST}:{PORT}/{DB_NAME}"
engine = create_engine(connection_string)

# 4. OVERWRITE the old messy table with your perfect new one
print("\nUploading clean data to MySQL...")
df.to_sql(name='customer_behaviour', con=engine, if_exists='replace', index=False)
print("Success! Your database table is now 100% clean and ready for queries.")
            
```
